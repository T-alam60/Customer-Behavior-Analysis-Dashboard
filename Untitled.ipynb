```python
import pandas as pd 
import numpy as np 
import matplotlib.pyplot as plt 
```

```python
df = pd.read_csv("customer_shopping_behavior.csv")
```

```python
df.isnull().sum()
```

```python
df["Review Rating"] = df.groupby("Category")["Review Rating"].transform(lambda x:x.fillna(x.median()))
```

```python
df.isnull().sum()
```

```python
df.columns = df.columns.str.lower()
df.columns = df.columns.str.replace(' ','_')
df = df.rename(columns={'purchase_amount_(usd)':'perchase_ammount'})
```

```python
df.columns
```

```python
labels = ['young-adult','adult','midle-aged','senior']
df['age_group'] = pd.qcut(df['age'],q=4,labels=labels)
```

```python
df[['age','age_group']]
```

```python
df['frequency_of_purchases'].unique()
```

```python
frequency_mapping = {
    'Fortnightly':14, 
    'Weekly':7, 
    'Annually':365, 
    'Quarterly':90, 
    'Bi-Weekly':14,
       'Monthly':30,
    'Every 3 Months':90
}
```

```python
df['frequency_of_purchases_days'] =df['frequency_of_purchases'].map(frequency_mapping)
```

```python
df[['frequency_of_purchases','frequency_of_purchases_days']].head()
```

```python
df.head()
```

```python
(df['promo_code_used'] == df['discount_applied']).all()
```

```python
df = df.drop(['promo_code_used','frequency_of_purchases_days'] , axis=1)
```

```python
df.head()
```

```python
df.groupby('gender')['perchase_ammount'].sum()
```

```python
df.groupby('gender')['perchase_ammount'].sum()

df.groupby('gender')['perchase_ammount'].sum().plot(kind='bar')
plt.xlabel('Gender')
plt.ylabel('Total Purchase Amount')
plt.title('Total Purchase Amount by Gender')
plt.xticks(rotation=0)
plt.show()
```

```python
avg_purchase = df['perchase_ammount'].mean()

q2 = df[(df['discount_applied'] == 'Yes') & 
        (df['perchase_ammount'] > avg_purchase)]
```

```python


# Q2
avg_purchase = df['perchase_ammount'].mean()
q2 = df[(df['discount_applied'] == 'Yes') & (df['perchase_ammount'] > avg_purchase)]

print("Q2: Customers who used discount and spent more than average")
print(q2[['customer_id', 'gender', 'item_purchased', 'perchase_ammount']])
print("Number of customers:", len(q2))

q2.groupby('gender')['perchase_ammount'].sum().plot(kind='bar', figsize=(7,4))
plt.title('Q2: Discounted Customers Spending Above Average')
plt.xlabel('Gender')
plt.ylabel('Total Purchase Amount')
plt.xticks(rotation=0)
plt.show()


# Q3
q3 = df.groupby('item_purchased')['review_rating'].mean().sort_values(ascending=False).head(5)

print("Q3: Top 5 Products with Highest Average Review Rating")
print(q3)

q3.plot(kind='bar', figsize=(7,4))
plt.title('Q3: Top 5 Products by Average Review Rating')
plt.xlabel('Product')
plt.ylabel('Average Review Rating')
plt.xticks(rotation=45)
plt.show()


# Q4
q4 = df[df['shipping_type'].isin(['Standard', 'Express'])].groupby('shipping_type')['perchase_ammount'].mean()

print("Q4: Average Purchase Amount - Standard vs Express Shipping")
print(q4)

q4.plot(kind='bar', figsize=(7,4))
plt.title('Q4: Standard vs Express Shipping')
plt.xlabel('Shipping Type')
plt.ylabel('Average Purchase Amount')
plt.xticks(rotation=0)
plt.show()


# Q5
q5 = df.groupby('subscription_status')['perchase_ammount'].agg(
    average_spend='mean',
    total_revenue='sum'
)

print("Q5: Subscribers vs Non-Subscribers")
print(q5)

q5['average_spend'].plot(kind='bar', figsize=(7,4))
plt.title('Q5: Average Spend - Subscribers vs Non-Subscribers')
plt.xlabel('Subscription Status')
plt.ylabel('Average Spend')
plt.xticks(rotation=0)
plt.show()

q5['total_revenue'].plot(kind='bar', figsize=(7,4))
plt.title('Q5: Total Revenue - Subscribers vs Non-Subscribers')
plt.xlabel('Subscription Status')
plt.ylabel('Total Revenue')
plt.xticks(rotation=0)
plt.show()


# Q6
q6 = df.groupby('item_purchased').agg(
    total_purchases=('item_purchased', 'count'),
    discounted_purchases=('discount_applied', lambda x: (x == 'Yes').sum())
)

q6['discount_percentage'] = (q6['discounted_purchases'] / q6['total_purchases']) * 100
q6 = q6.sort_values('discount_percentage', ascending=False).head(5)

print("Q6: Top 5 Products with Highest Percentage of Discounted Purchases")
print(q6)

q6['discount_percentage'].plot(kind='bar', figsize=(7,4))
plt.title('Q6: Top 5 Products by Discount Percentage')
plt.xlabel('Product')
plt.ylabel('Discount Percentage (%)')
plt.xticks(rotation=45)
plt.show()
```
